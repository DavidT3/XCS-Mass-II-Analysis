# New SDSSRM-XCS Relations: $M_{\rm{tot}}$-$T_{\rm{X}}$

This section of the project fits scaling relations to the measurements of properties made in the first paper of this series [(Turner et al. 2024)](https://ui.adsabs.harvard.edu/abs/2025MNRAS.tmp...10T/abstract). Masses are from the 'hydrostatic equilibrium' method, fully explained in the initial paper.

Here we focus on scaling relations between the total mass of galaxy clusters and groups ($M_{\rm{tot}}$, or $M^{\rm{tot}}_{\Delta}$ within an overdensity radius $R_{\Delta}$) and the cluster/group global X-ray temperature ($T_{\rm{X}}$, or $T_{\rm{X},\Delta}$ within $R_{\Delta}$). If present in a property name, 'ce' means 'core-excised', normally 0.15$R_{500}$.

Relation fits are performed using the XGA interface to the R scaling-relation fitting package LIRA, see the paper for full details.

## Loading data files

Important considerations for this dataset:
* The X-ray cluster sample was originally assembled and analysed by [Giles et al. (2022)](https://ui.adsabs.harvard.edu/abs/2022MNRAS.516.3878G/abstract) - when we say 'SDSSRM-XCS' we are specifically referring to the volume limited ($0.1<z<0.35$) sample with fractional temperature errors of less than 25% - contains 150 clusters.
* SDSSRM-XCS is ultimately an *optically selected* sample however, as the original X-ray work (see above) found X-ray counterparts to clusters selected from the redMaPPer catalog of galaxy clusters produced from SDSS photometry [(Rykoff et al. 2014)](http://adsabs.harvard.edu/abs/2014ApJ...785..104R).
* These galaxy clusters were then further analysed in the first paper of this series [(Turner et al. 2024)](https://ui.adsabs.harvard.edu/abs/2025MNRAS.tmp...10T/abstract), measuring radial profiles of density and temperature that were used to calculate hydrostatic and gas masses.

## Fitting scaling relations

### $M^{\rm{tot}}_{500}$-$T_{\rm{X,500}}$

```python
# Setting up property variables in astropy quantity objects
mtot500 = Quantity(sdssxcs_samp[['Mhy500_wraderr', 'Mhy500_wraderr-', 'Mhy500_wraderr+']].values*1e+14, 'Msun')\
    *sdssxcs_samp['E'].values[..., None]
tx500 = Quantity(sdssxcs_samp[['Tx_500', 'Tx_500-','Tx_500+']].values, 'keV')

mtot500_tx500 = scaling_relation_lira(mtot500[:, 0], mtot500[:, 1:], tx500[:, 0], tx500[:, 1:], m_norm, tx_norm, 
                                      y_name=r"$E(z)M^{\rm{tot}}_{500}$", x_name=r"$T_{\rm{X,500}}$", 
                                      dim_hubb_ind=1, point_names=sdssxcs_samp['name'].values)
mtot500_tx500.model_colour = 'tab:cyan'
mtot500_tx500.author = 'Turner et al.'
mtot500_tx500.year = 2025
mtot500_tx500.name = r'Turner et al. $E(z)M^{\rm{tot}}_{500}$-$T_{\rm{X,500}}$'
mtot500_tx500.save("../../outputs/scaling_relations/sdssrm-xcs_new/mtot-tx/turner2025_mtot500_tx500.xgarel")
```

```python
print("Slope of {v:.3f} ± {e:.3f}\n".format(v=mtot500_tx500.pars[0][0], e=mtot500_tx500.pars[0][1]))
print("Normalisation of {v:.3f} ± {e:.3f}\n".format(v=mtot500_tx500.pars[1][0], e=mtot500_tx500.pars[1][1]))
print("Scatter of {v:.3f} ± {e:.3f}".format(v=mtot500_tx500.scatter_par[0], e=mtot500_tx500.scatter_par[1]))
```

## Limited-$T_{\rm{X}}$-range scaling relations

In order to properly the scatter in our mass-temperature dataset to the scatter in the Lovisari et al. (2020) dataset, we will select a subset of the SDSSRM-XCS sample and fit a scaling relation. The idea being that we will select SDSSRM-XCS clusters to match the temperature range of the Lovisari et al. systems, as we consider that the lower mass systems we probe, but that are not represented in the Lovisari set, may be making an outsized contribution to the scatter.

```python
lov_low_tx = Quantity(3.9, 'keV')
```

### $M^{\rm{tot}}_{500}$-$T_{\rm{X,500}}$

```python
# Setting up property variables in astropy quantity objects - THEN CUTTING CLUSTERS BELOW THE TEMPERATURE LIMIT
tx500 = Quantity(sdssxcs_samp[['Tx_500', 'Tx_500-','Tx_500+']].values, 'keV')
lim_tx500_msk = (tx500[:, 0] > lov_low_tx)

# The masked temperatures and mass data points
tx500 = tx500[lim_tx500_msk, :]
mtot500 = (Quantity(sdssxcs_samp[['Mhy500_wraderr', 'Mhy500_wraderr-', 'Mhy500_wraderr+']].values*1e+14, 'Msun')
               *sdssxcs_samp['E'].values[..., None])[lim_tx500_msk, :]

mtot500_tx500_txlim = scaling_relation_lira(mtot500[:, 0], mtot500[:, 1:], tx500[:, 0], tx500[:, 1:], m_norm, tx_norm, 
                                            y_name=r"$E(z)M^{\rm{tot}}_{500}$", x_name=r"$T_{\rm{X,500}}$", 
                                            dim_hubb_ind=1, point_names=sdssxcs_samp['name'].values[lim_tx500_msk])
mtot500_tx500_txlim.model_colour = 'cadetblue'
mtot500_tx500_txlim.author = 'Turner et al.'
mtot500_tx500_txlim.year = 2025
mtot500_tx500_txlim.name = r'Turner et al. $E(z)M^{\rm{tot}}_{500}$-$T_{\rm{X,500}}$ [$T_{\rm{X}}>3.9$ keV]'
mtot500_tx500_txlim.save("../../outputs/scaling_relations/sdssrm-xcs_new/mtot-tx/turner2025_mtot500_tx500_txgt3_9keV.xgarel")
```

```python
print("Slope of {v:.3f} ± {e:.3f}\n".format(v=mtot500_tx500_txlim.pars[0][0], e=mtot500_tx500_txlim.pars[0][1]))
print("Normalisation of {v:.3f} ± {e:.3f}\n".format(v=mtot500_tx500_txlim.pars[1][0], e=mtot500_tx500_txlim.pars[1][1]))
print("Scatter of {v:.3f} ± {e:.3f}".format(v=mtot500_tx500_txlim.scatter_par[0], e=mtot500_tx500_txlim.scatter_par[1]))
```

```python
mtot500_tx500_txlim.view(figsize=(7.3, 7), y_lims=Quantity([1e+14, 4e+15], 'Msun'), label_points=False)
```
