# New SDSSRM-XCS Relations: $M_{\rm{tot}}$-$\lambda$

This section of the project fits scaling relations to the measurements of properties made in the first paper of this series [(Turner et al. 2024)](https://ui.adsabs.harvard.edu/abs/2025MNRAS.tmp...10T/abstract). Masses are from the 'hydrostatic equilibrium' method, fully explained in the initial paper.

Here we focus on scaling relations between the total mass of galaxy clusters and groups ($M_{\rm{tot}}$, or $M^{\rm{tot}}_{\Delta}$ within an overdensity radius $R_{\Delta}$) and the cluster/group richness ($\lambda$; a probabilistic measure of the number of galaxies in a cluster/group). We may also refer to richness as $\lambda_{\rm{RM}}$, to indicate that the SDSSRM-XCS richnesses were measured by the redMaPPer tool.

Relation fits are performed using the XGA interface to the R scaling-relation fitting package LIRA, see the paper for full details.

## Loading data files

Important considerations for this dataset:
* The X-ray cluster sample was originally assembled and analysed by [Giles et al. (2022)](https://ui.adsabs.harvard.edu/abs/2022MNRAS.516.3878G/abstract) - when we say 'SDSSRM-XCS' we are specifically referring to the volume limited ($0.1<z<0.35$) sample with fractional temperature errors of less than 25% - contains 150 clusters.
* SDSSRM-XCS is ultimately an *optically selected* sample however, as the original X-ray work (see above) found X-ray counterparts to clusters selected from the redMaPPer catalog of galaxy clusters produced from SDSS photometry [(Rykoff et al. 2014)](http://adsabs.harvard.edu/abs/2014ApJ...785..104R).
* These galaxy clusters were then further analysed in the first paper of this series [(Turner et al. 2024)](https://ui.adsabs.harvard.edu/abs/2025MNRAS.tmp...10T/abstract), measuring radial profiles of density and temperature that were used to calculate hydrostatic and gas masses.

## Fitting scaling relations

### $M^{\rm{tot}}_{500}$-$\lambda_{\rm{RM}}$

```python
# Setting up property variables in astropy quantity objects
mtot500 = Quantity(sdssxcs_samp[['Mhy500_wraderr', 'Mhy500_wraderr-', 'Mhy500_wraderr+']].values*1e+14, 'Msun')
rich = Quantity(sdssxcs_samp[['richness', 'richness_err','richness_err']].values, '')

mtot500_rich = scaling_relation_lira(mtot500[:, 0], mtot500[:, 1:], rich[:, 0], rich[:, 1:], m_norm, rich_norm, 
                                     y_name=r"$M^{\rm{tot}}_{500}$", x_name=r"$\lambda_{\rm{RM}}$", 
                                     point_names=sdssxcs_samp['name'].values)
mtot500_rich.model_colour = 'goldenrod'
mtot500_rich.author = 'Turner et al.'
mtot500_rich.year = 2025
mtot500_rich.name = r'Turner et al. $M^{\rm{tot}}_{500}$-$\lambda_{\rm{RM}}$'
mtot500_rich.save("../../outputs/scaling_relations/sdssrm-xcs_new/mtot-richness/turner2025_mtot500_richness.xgarel")
```

```python
print("Slope of {v:.3f} ± {e:.3f}\n".format(v=mtot500_rich.pars[0][0], e=mtot500_rich.pars[0][1]))
print("Normalisation of {v:.3f} ± {e:.3f}\n".format(v=mtot500_rich.pars[1][0], e=mtot500_rich.pars[1][1]))
print("Scatter of {v:.3f} ± {e:.3f}".format(v=mtot500_rich.scatter_par[0], e=mtot500_rich.scatter_par[1]))
```

```python
mtot500_rich.view(figsize=(7.3, 7), y_lims=Quantity([1e+13, 4e+15], 'Msun'), 
                  save_path="../../outputs/figures/scaling_relations/sdssrm-xcs_new/mtot-richness/turner2025_mtot500_richness.pdf")
```

## Limited-$\lambda_\rm{RM}$-range scaling relations

In order to properly the scatter in our mass-richness dataset to the scatter in the Mulroy et al. (2019) dataset, we will select a subset of the SDSSRM-XCS sample and fit a scaling relation. The idea being that we will select SDSSRM-XCS clusters to match the richness range of the Mulroy et al. LoCuSS systems, as we consider that the lower mass systems we probe, but that are not represented in the Mulroy set, may be making an outsized contribution to the scatter.

```python
mul_low_rich = Quantity(45, '')
```

### $M^{\rm{tot}}_{500}$-$\lambda_{\rm{RM}}$

```python
rich = Quantity(sdssxcs_samp[['richness', 'richness_err','richness_err']].values, '')
lim_rich_msk = (rich[:, 0] > mul_low_rich)

# The masked richness and mass data points
rich = rich[lim_rich_msk, :]
mtot500 = Quantity(sdssxcs_samp[['Mhy500_wraderr', 'Mhy500_wraderr-', 'Mhy500_wraderr+']].values*1e+14, 'Msun')[lim_rich_msk, :]

mtot500_rich_richlim = scaling_relation_lira(mtot500[:, 0], mtot500[:, 1:], rich[:, 0], rich[:, 1:], m_norm, rich_norm, 
                                             y_name=r"$M^{\rm{tot}}_{500}$", x_name=r"$\lambda_{\rm{RM}}$", 
                                             point_names=sdssxcs_samp['name'].values[lim_rich_msk])
mtot500_rich_richlim.model_colour = 'bisque'
mtot500_rich_richlim.author = 'Turner et al.'
mtot500_rich_richlim.year = 2025
mtot500_rich_richlim.name = r'Turner et al. $M^{\rm{tot}}_{500}$-$\lambda_{\rm{RM}}$ [$\lambda_{\rm{RM}}>45$]'
mtot500_rich_richlim.save("../../outputs/scaling_relations/sdssrm-xcs_new/mtot-richness/turner2025_mtot500_richness_richgt45.xgarel")
```

```python
print("Slope of {v:.3f} ± {e:.3f}\n".format(v=mtot500_rich_richlim.pars[0][0], e=mtot500_rich_richlim.pars[0][1]))
print("Normalisation of {v:.3f} ± {e:.3f}\n".format(v=mtot500_rich_richlim.pars[1][0], e=mtot500_rich_richlim.pars[1][1]))
print("Scatter of {v:.3f} ± {e:.3f}".format(v=mtot500_rich_richlim.scatter_par[0], e=mtot500_rich_richlim.scatter_par[1]))
```
